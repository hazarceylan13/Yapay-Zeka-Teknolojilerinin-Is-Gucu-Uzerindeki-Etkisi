## BÖLÜM 1 (MP1): Araştırma Problemi ve Hipotezler

### 1.1. Problem Tanımı
Teknolojik gelişmeler tarih boyunca iş yapış şekillerini değiştirmiştir. Ancak Yapay Zeka (YZ) devrimi, sadece kas gücüne dayalı işleri değil, bilişsel yetenek gerektiren işleri de tehdit etmektedir. Temel sorun, bu dönüşümün istihdamı azaltıp azaltmayacağı veya hangi yetkinlikleri öne çıkaracağı belirsizliğidir.

### 1.2. Araştırmanın Değişkenleri
Çalışmanın analitik modeli aşağıdaki bağımlı ve bağımsız değişkenler üzerine kurgulanmıştır:

* **Bağımsız Değişkenler (Independent Variables):**
    1.  **Otomasyon Potansiyeli (%):** Mevcut teknoloji ile bir mesleğin görevlerinin makineye devredilme oranı.
    2.  **YZ Maruziyet Skoru (1-100):** Bir mesleğin üretken yapay zeka (LLM vb.) araçlarıyla etkileşim düzeyi.
    3.  **Sektör:** Mesleğin ait olduğu endüstri.

* **Bağımlı Değişken (Dependent Variable):**
    1.  **Gelecek Talep Değişimi (%):** 2023-2027 yılları arasında o mesleğe yönelik istihdam talebindeki beklenen artış veya azalış oranı.

### 1.3. Hipotezler
Belirlenen hipotezler şunlardır:

* **H1:** Farklı sektörlerde yapay zeka ve otomasyon düzeyi yükseldikçe; bu teknolojilere adapte olamayan ve düşük beceri gerektiren meslek gruplarında istihdam oranı düşme eğilimi gösterir.
* **H0 (Boş Hipotez):** Yapay zeka teknolojilerinin iş gücü üzerinde anlamlı bir etkisi bulunmamaktadır.

## BÖLÜM 2 (MP2): Literatür Taraması ve Araştırma Boşluğu

### 2.1. Literatür Özeti
Bu çalışmanın teorik altyapısı, yapay zeka ve istihdam ilişkisini inceleyen güncel akademik çalışmaların, küresel raporların ve makalelerin sentezlenmesiyle oluşturulmuştur. İncelenen kapsamlı literatür aşağıda kronolojik olarak sunulmuştur:

1.  **Brynjolfsson, Chandar, Chen (2025) - "Impact of AI on Youth Employment"**
     * **Yöntem:** ABD bordro verileri kullanılarak yapılan yüksek frekanslı veri analizi.
     * **Temel Sonuç:** Yapay zekaya maruz kalan, özellikle 22–25 yaş aralığındaki genç çalışanlarda istihdamda **%16'lık bir azalma** gözlemlenmiştir. Bu durum, giriş seviyesindeki pozisyonların YZ tarafından ikame edildiğini göstermektedir.

2.  **World Economic Forum (2025) - Küresel Dönüşüm Tahmini**
     * **Yöntem:** Küresel verilerle yapılan tahminsel modelleme.
     * **Temel Sonuç:** Yapay zeka ve otomasyonun iş gücü piyasasında yarattığı dönüşüm sonucunda, 2030 yılına kadar dünya genelinde 170 milyon yeni iş yaratılacağı, buna karşılık 92 milyon işin yer değiştireceği (kaybolacağı) öngörülmektedir. Bu senaryoya göre, teknolojik dönüşümün küresel istihdama net etkisi yaklaşık +78 milyon yeni iş ile pozitif yönde olacaktır.

3.  **World Economic Forum (WEF) - Future of Jobs Report 2023**
     * **Bulgu:** Önümüzdeki 5 yıl içinde küresel iş gücü piyasasında %23 oranında bir "churn" (mesleki çalkalanma) beklenmektedir. Rapor, 69 milyon yeni iş yaratılacağını, ancak 83 milyon işin yok olacağını (net 14 milyon düşüş) öngörmektedir.
     * **Etki:** İdari işler, sekreterlik ve fabrika işçiliğinde düşüş öngörülürken; "Büyük Veri Uzmanlığı", "Makine Öğrenimi Mühendisliği" ve "Siber Güvenlik" alanlarında en hızlı istihdam artışı beklenmektedir.

4.  **OpenAI & Stanford HAI - "GPTs are GPTs" Araştırması (2023)**
     * **Bulgu:** Büyük Dil Modellerinin (LLM), ABD ekonomisindeki işlerin %80'inin görevlerinin en az %10'unu otomatize edebileceği tespit edilmiştir.
     * **Etki:** Araştırma, yaygın kanının aksine, **yüksek gelirli ve yüksek eğitimli** mesleklerin (yazılımcılar, veri analistleri, yazarlar) YZ'ye maruz kalma riskinin (Exposure), fiziksel işlere göre daha yüksek olduğunu kanıtlamıştır.

5.  **Goldman Sachs - The Potentially Large Effects of AI on Economic Growth (2023)**
     * **Bulgu:** Üretken yapay zekanın (Generative AI) küresel GSYİH'yı yıllık %7 oranında artırabileceği (yaklaşık 7 trilyon dolar) öngörülmüştür.
     * **Etki:** Rapor, otomasyonun 300 milyon tam zamanlı işi etkileyebileceğini belirtmekle birlikte, bu dönüşümün verimlilik artışı sayesinde yeni iş kolları yaratacağını savunmaktadır.

6.  **McKinsey Global Institute - The Future of Work in America (2023)**
     * **Bulgu:** Mevcut teknolojilerle, iş aktivitelerinin %50'sinin otomatize edilebileceği hesaplanmıştır.
     * **Etki:** 2030 yılına kadar ABD'de 12 milyon insanın meslek değiştirmek zorunda kalacağı tahmin edilmektedir. Rapor, fiziksel ve rutin işlerden; bilişsel ve sosyal beceri gerektiren işlere doğru büyük bir kayma olacağını vurgulamaktadır.

7.  **Noy, S., & Zhang, W. (2023). "Experimental Evidence on the Productivity Effects of Generative Artificial Intelligence". Science.**
     * **Yöntem:** Çin üretim sektörü verileri üzerinde makine öğrenimi tabanlı tahmin modeli.
     * **Temel Sonuç:** Yapay zeka uygulamaları üretimde verimliliği belirgin şekilde yükseltmiş, fakat buna karşılık düşük vasıflı işlerde çalışan sayısında azalma görülmüştür.

8.  **OECD (2023) - İstihdam Görünümü**
     * **Yöntem:** 32 ülkenin istihdam verileriyle yapılan ekonomik analiz ve politika incelemesi.
     * **Temel Sonuç:** Yapay zeka, mevcut meslekleri dönüştürerek beceri talebini değiştirmekte ve tamamen yeni meslek gruplarının ortaya çıkmasına yol açmaktadır.

9. **Webb (2020) - "The Impact of Artificial Intelligence on the Labor Market"**
     * **Yöntem:** Meslek tanımları ve görev verileri üzerinden yapılan istihdam analizi.
     * **Temel Sonuç:** Yapay zeka, önceki otomasyon dalgalarından farklı olarak sadece fiziksel işleri değil, bilişsel yetenek gerektiren işleri de etkilemektedir. Ancak tekrar eden görevler yüksek oranda etkilenirken, karmaşık problem çözme gerektiren işler daha dirençlidir.

10.  **Zhou, Y., & Tyers, R. (2019). "Automation and Inequality in China."**
     * **Yöntem:** Anket verilerine dayalı istatistiksel analiz.
     * **Temel Sonuç:** Yapay zeka, özellikle tekrarlayan ve basit görevleri yerine getirerek düşük vasıflı işler üzerinde olumsuz bir etki yaratmaktadır. Rutin işlerdeki insan emeğine olan ihtiyaç azalma eğilimindedir.


### 2.2. Araştırma Boşluğu (Research Gap)
Literatür taraması sonucunda şu eksiklik tespit edilmiştir:

Mevcut çalışmalar, yapay zeka ve otomasyonun iş gücü üzerindeki etkilerini çeşitli ülkeler ve sektörler için incelemiştir; ancak **meslek bazında**, farklı beceri seviyeleri ve çalışma koşullarına göre etkilerin karşılaştırmalı olarak analiz edilmesi sınırlı kalmıştır. Ayrıca, yapay zeka adaptasyonu ile çalışan verimliliği ve yeni iş fırsatları arasındaki ilişkiyi ölçen kapsamlı çalışmalar halen azdır.

## BÖLÜM 4 (MP4): Veri Analizi ve Görselleştirme

Bu bölümde Nicel Veri Seti kullanılarak grafikler oluşturulmuş ve analizler yapılmıştır.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

df_nicel = pd.read_csv('data_mp3_2.csv', sep=';')
df_nicel.index = df_nicel.index + 1

df = df_nicel.copy()

print("--- İstatistiksel Özet ---")
print(df.describe().to_string())
```

```python
# Korelasyon Matrisi
corr_matrix = df.select_dtypes(include=['number']).corr()

print("--- Korelasyon Değerleri ---")
print(corr_matrix.to_string())

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=1)
plt.title('Değişkenler Arasındaki İlişki (Korelasyon)', fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(14, 6))
sns.histplot(df['Otomasyon Potansiyeli Yüzdesi'], bins=10, kde=True, color='skyblue', edgecolor='black')

plt.title('Mesleklerin Otomasyon Riski Dağılımı', fontsize=14)
plt.xlabel('Otomasyon Potansiyeli (%)')
plt.ylabel('Meslek Sayısı')
plt.show()
```

```python
plt.figure(figsize=(14, 9))

sns.scatterplot(data=df, 
                x='Yapay Zekaya Maruz Kalma Skoru (1-100)', 
                y='Otomasyon Potansiyeli Yüzdesi', 
                hue='Sektör', 
                palette='tab10',   
                s=200,             
                alpha=0.9, 
                edgecolor='black',
                linewidth=1.2)

plt.title('YZ Maruziyeti vs. Otomasyon Riski (Sektörel Bakış)', fontsize=16)
plt.xlabel('Yapay Zekaya Maruz Kalma Skoru (1-100)', fontsize=12)
plt.ylabel('Otomasyon Potansiyeli (%)', fontsize=12)

plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', title='Sektörler')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(12, 8))

talep_sutunu = 'Gelecek Talep Değişimi Tahmini (yüzdelik değişim)'
meslek_sutunu = 'Meslek Ünvanı'

df_sirali = df.sort_values(talep_sutunu, ascending=False)

top_bottom = pd.concat([df_sirali.head(7), df_sirali.tail(7)])

colors = {meslek: 'green' if talep > 0 else 'red' 
          for meslek, talep in zip(top_bottom[meslek_sutunu], top_bottom[talep_sutunu])}

sns.barplot(data=top_bottom, 
            x=talep_sutunu, 
            y=meslek_sutunu, 
            hue=meslek_sutunu, 
            palette=colors,    
            legend=False)     

plt.title('Gelecekte Talebi En Çok Artacak ve Azalacak Meslekler', fontsize=15)
plt.xlabel('Talep Değişimi (%)')
plt.ylabel('Meslekler')
plt.axvline(0, color='black', linewidth=1) 
plt.show()
```

## BÖLÜM 5 (MP5): RAPORLAMA VE SONUÇ

### 5.1. Temel Bulgular ve Tartışma
Proje kapsamında gerçekleştirilen veri analizi ve literatür taraması sonucunda, yapay zeka ve istihdam arasındaki ilişkiye dair aşağıdaki kritik bulgulara ulaşılmıştır:

**1.  Otomasyon ve Talep Arasındaki Ters İlişki (Negatif Korelasyon):**
    Analiz sonuçlarımız, bir mesleğin otomasyon potansiyeli ile gelecekteki talep beklentisi arasında **güçlü bir negatif ilişki (-0.85 Korelasyon)** olduğuna işaret etmektedir. Bu bulgu, iş süreçlerinin makineleşmeye uygun hale geldikçe, insan iş gücüne duyulan ihtiyacın **azalma eğilimi gösterdiği** yönündeki literatür bilgisini desteklemektedir. Özellikle **idari işler ve rutin veri işleme** görevleri bu bağlamda en yüksek risk grubunu oluşturmaktadır.

**2.  Sektörel Direnç ve Ayrışma:**
    Veriler incelendiğinde, her sektörün YZ devriminden aynı düzeyde etkilenmediği gözlemlenmiştir. **Teknoloji ve Yazılım** sektörü, yüksek YZ maruziyetine (High Exposure) sahip olmasına rağmen, otomasyon riskine karşı daha dirençli bir yapı sergilemektedir. Buna karşın, **Hizmet ve Lojistik** sektörlerinde yüksek maruziyet oranlarının, istihdam kaybı riskiyle **paralellik gösterdiği** izlenmiştir.

**3.  Yapay Zekanın Çift Yönlü Etkisi (İkame vs. Tamamlayıcılık):**
    Elde edilen veriler, yapay zekanın iş gücü piyasasında iki farklı rol üstlendiğini göstermektedir:
    * **İkame Etkisi (Substitution):** Düşük beceri ve tekrar gerektiren işlerde (Örn: Veri Giriş Operatörü), YZ teknolojilerinin insan emeğini ikame etme potansiyeli yüksektir.
    * **Tamamlayıcılık Etkisi (Complementarity):** Bilişsel ve yaratıcı yetkinlik gerektiren işlerde (Örn: Veri Bilimci), YZ'nin verimliliği artırıcı bir araç olarak konumlandığı görülmektedir.

**4.  Yetkinlik Dönüşümü Zorunluluğu (Reskilling):**
    Analizler, geleceğin iş dünyasında rutin görevlerin azalacağını; bunun yerine **analitik düşünme, karmaşık problem çözme ve YZ okuryazarlığı** gibi yetkinliklerin temel kriter haline geleceğini öngörmektedir.
